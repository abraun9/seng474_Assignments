```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from mpl_toolkits.mplot3d import Axes3D

np.random.seed(15)
```

```python
path_dataset1 = r'C:\Users\abrau\uvic\seng474\seng474_Assignments\assignment3\files\dataset1.csv'
path_dataset2 = r'C:\Users\abrau\uvic\seng474\seng474_Assignments\assignment3\files\dataset2.csv'

dataset1 = pd.read_csv(path_dataset1, header=None)
dataset2 = pd.read_csv(path_dataset2, header=None)
```

```python
colours = np.linspace(0, 5, len(dataset1))

plt.figure(figsize=(10,6))
plt.scatter(dataset1[0], dataset1[1], c=colours, cmap='plasma', label='Data Set 1')
plt.xlabel('X-axis (column 0)')
plt.ylabel('Y-axis (column 1)')
plt.title('Scatter Plot Data Set 1')
plt.legend()
plt.show()
```

```python
colours = np.linspace(0, 5, len(dataset2))

fig = plt.figure(figsize=(10,8))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(dataset2[0], dataset2[1], dataset2[2], c=colours, cmap='plasma', marker='o', label='3D Data')

ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
ax.set_title('3D Scatter Plot of dataset2')

ax.legend()

plt.show()
```

```python
import plotly.express as px
import pandas as pd

# Create a DataFrame for the 3D data
df = pd.DataFrame({
    'X': dataset2[0],
    'Y': dataset2[1],
    'Z': dataset2[2],
    'Color': colours  # Add the color gradient as a column
})

# Create an interactive 3D scatter plot
fig = px.scatter_3d(df, x='X', y='Y', z='Z', color='Color', color_continuous_scale='viridis')

# Update layout
fig.update_layout(
    title='3D Scatter Plot with Progressive Coloring',
    scene=dict(
        xaxis_title='X-axis',
        yaxis_title='Y-axis',
        zaxis_title='Z-axis'
    )
)

# Show the plot
fig.show()
```
